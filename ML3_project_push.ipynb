```
import  numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import  accuracy_score
```

```
# loading the dataset to a pandas Dataframe
wine_dataset = pd.read_csv('/content/winequality-red.csv')
```

```
wine_dataset.shape
```

```
# first 5 rows of dataset
wine_dataset.head()
```

```
# checking missing value
wine_dataset.isnull().sum()
```

```
# statistical measure of dataset
wine_dataset.describe()
```

```
#  number of values for each  quality
sns.catplot(x='quality', data = wine_dataset, kind = 'count')
```

```
# volatile acidity vs quality
plot = plt.figure(figsize=(5,5))
sns.barplot(x='quality', y = 'volatile acidity', data = wine_dataset)
```

```
# citric acid vs Quality
plot = plt.figure(figsize=(5,5))
sns.barplot(x='quality', y = 'citric acid', data = wine_dataset, color = 'g')
plt.show()
```

1.Positive Correlation
2.Negative Correlation

```
coorelation = wine_dataset.corr()
```

```
# constucting a sns.heatmap to understand the correaltion between the coloumn
plt.figure(figsize=(10,10))
sns.heatmap(coorelation,cbar=True,square=True,fmt = '.1f',annot=True,annot_kws={'size': 8},cmap='Blues')
```

```
# separete the data and label
X = wine_dataset.drop('quality',axis=1)
```

```
print(X)
```

```
Y = wine_dataset['quality'].apply(lambda y_value : 1 if y_value >= 7 else 0)
```

```
print(Y)
```

```
# Train & Test
X_train ,X_test,Y_train ,Y_test = train_test_split(X,Y, test_size=0.2,random_state=3 )
```

```
print(Y.shape,Y_train.shape,Y_test.shape)
```

```
# random forst  classifir
model = RandomForestClassifier()
```

```
model.fit(X_train,Y_train)
```

```
Training_data_predions = model.predict(X_train)
```

```
check_score = accuracy_score(Y_train,Training_data_predions)
print('Accuracy :',check_score)
```

```
plt.scatter(Y_train,Training_data_predions)
plt.show()
```

```
# Accuracy Scores
X_test_prediction = model.predict(X_test)
test_data_accuracy= accuracy_score(X_test_prediction,Y_test)
```

```
print('Accuracy :',test_data_accuracy)
```

```
plt.scatter(X_test_prediction,Y_test)
plt.show()
```

```
input_data = (7.5,0.5,0.36,6.1,0.071,17.0,102.0,0.9978,3.35,0.8,10.5)
# chaninh the input data to numpy arrary
input_data_as_numpy_array = np.asarray(input_data)
# reshape the data we are prediction the lable for only on intstance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)
prediction = model.predict(input_data_reshaped)
print(prediction)
if (prediction[0]==1):
  print('Good Quality Wine')
else:
  print('Bad Quality Wine')
```
